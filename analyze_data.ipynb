```python
import json
import pandas as pd
from pathlib import Path

data_dir = Path("data/all")

dfs = []
for path in sorted(data_dir.glob("*.json")):
    if path.stat().st_size == 0:
        continue
    season = path.stem.replace("_", "-")  # "17_18" → "17-18"
    with open(path) as f:
        records = json.load(f)
    df = pd.DataFrame(records)
    df["season"] = season
    dfs.append(df)

combined_all = pd.concat(dfs, ignore_index=True)
print(combined_all.shape)
combined_all.head()
```

```python
combined_all.to_json("data/combined/combined_all.json", index=False)
```
